***INTEGER DIVISON WITH SCALAR***

Integer Division: Division in which the fractional part
(remainder) is discarded

```r
2 %/% 2
```

```r
5 %/% 2
```

```r
7 %/% 3
```

```r
c(2,3,5,7) %/% 2
```

```r
c(2,3,5,7) %/% c(2,3)
```

***MODULO DIVISION (X MOD Y) WITH SCALARS***

Modulo Division: modulo operation finds the remainder
after division of one number by another

```r
2 %% 2
```

```r
3 %% 2
```

```r
7 %% 3
```

```r
7 %% 4
```

```r
c(2,3,5,7) %% 2
```

```r
c(2,3,5,7) %% c(2,3)
```

```r
c(2,3,5) %% c(2,3)
```
